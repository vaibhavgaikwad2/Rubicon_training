```python
class Animal:
    def __init__(self,name):
        self.name=name

    def eat(self):
        print(f"Animal is eating who's name is {self.name}")

class Dog(Animal):
    def bark(self):
        print("Dog is Barking")


d=Dog("abc")
```

```python
d.eat()
```

```python
d.bark()
```

```python
class Animal:
    def sound(self):
        print("Animal Sound")

class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")


animals=[Dog(),Cat()]

for animal in animals:
    animal.sound()
```

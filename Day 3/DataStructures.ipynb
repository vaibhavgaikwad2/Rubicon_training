```python
# Freq counter using dict
lst=[1,1,2,3,4,5,6,7,7,7,8,88,8,]
freq={}

for i in lst:
    if i in freq:
        freq[i]+=1

    else:
        freq[i]=1


for key, value in freq.items():
    print(f"{key}:{value}")
```

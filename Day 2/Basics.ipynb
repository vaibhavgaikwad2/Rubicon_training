```python
string="apple angle ball bottle wall"
result=[]
freq={}
for wrd in string.split():
    res=wrd[0]+wrd[-1]
    result.append(res)

for item in result:
    if item in freq:
        freq[item]+=1

    else:
        freq[item]=1

for item in freq.items():
    print(item)

#print(result)
```
